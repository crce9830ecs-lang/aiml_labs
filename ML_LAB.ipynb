```python
#creating arrays
import numpy as np
arr = np.array([4,5,6,3,2,1])
print(arr)

b = np.arange(1,11) #give a range and then function will create an array
print(b)

c = np.linspace(0,1,5)
print(c)    #0: The starting point of the sequence.1: The end point of the sequence (included by default).5: The total number of elements to generate.
```

```python
# matrix operation
a =np.array(([1,2],
             [3,4]))
b =np.array(([2,2],
             [4,4]))
print(a+b)
print(a-b)
print(a*b) #element wise multiplication
print(a@b) # matix multiplication
```

```python
#assignment
# weather data of july days in udupi,
# - find hottest and clodest day
# - ang temp
# - dates above 35 deg cel
import numpy as np

weather = np.array([
    [1, 26], [2, 28], [3, 28], [4, 29], [5, 30],
    [6, 30], [7, 30], [8, 30], [9, 31], [10, 30],
    [11, 30], [12, 30], [13, 30], [14, 31], [15, 31],
    [16, 30], [17, 27], [18, 29], [19, 29], [20, 27],
    [21, 29], [22, 26], [23, 29], [24, 30], [25, 29],
    [26, 29], [27, 30], [28, 28], [29, 29], [30, 30],
    [31, 30]
])

index = np.argmin(weather[:,1])
date = weather[index, 0]
temp = weather[index, 1]
print(f"hottest day is on {date} with a temperature of {temp}")

index = np.argmax(weather[:,1])
date = weather[index, 0]
temp = weather[index, 1]
print(f"coldest day is on {date} with a temperature of {temp}")

avg = np.mean(weather[:,1])
print(f"the avg temp is {avg}")

if ((weather[:,1]) > 35).all():
  print((weather[:,1]))
```

```python
# titanic dataset kaggle/github
# find total no of passengers
# avg age of surviors and non surviors
# passengers older than 60
# avg money for each passenger class

import kagglehub
import pandas as pd
import os
# Download latest version
path = kagglehub.dataset_download("yasserh/titanic-dataset")

print("Path to dataset files:", path)

csv_filename = os.listdir(path)[0]  # Gets the first file in the downloaded folder
file_path = os.path.join(path, csv_filename)
df = pd.read_csv(file_path)

df.head(10)

#total no of passengers
total_pass = len(df)
print(f"total no of passengers: {total_pass}")

# avg age of surving passengers
sur = df[df["Survived"]==1]["Age"].mean()
print(f"avg age of surving passengers: {sur}")

#avg age of non surving passengers
non_sur = df[df["Survived"] == 0]["Age"].mean()
print(f"avg age if non surving passengers: {non_sur}")

older = len(df[df["Age"] > 60])
print(f"no of passengers old than 60: {older}")

print(df.groupby("Pclass").agg({"Fare": "mean"}))
```
